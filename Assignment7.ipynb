```sql
--1. Find all laptops that cost more than the average laptop price.
select * from Laptop_sales 
where price > (select avg(price) from Laptop_sales);
```

```sql
--2. Find the most expensive laptop.
SELECT *
FROM Laptop_Sales
WHERE Price =
(
    SELECT MAX(Price)
    FROM Laptop_Sales
);
```

```sql
--3. Find the cheapest laptop.
SELECT *
FROM Laptop_Sales
WHERE Price =
(
    SELECT MIN(Price)
    FROM Laptop_Sales
);
```

```sql
--4. Find laptops sold in the same store as the MacBook Air.
SELECT *
FROM Laptop_Sales
WHERE Store =
(
    SELECT Store
    FROM Laptop_Sales
    WHERE Model = 'MacBook Air'
);
```

```sql
--5. Find laptops whose quantity sold is greater than the average quantity sold.
 SELECT *
FROM Laptop_Sales
WHERE QuantitySold >
(
    SELECT AVG(QuantitySold)
    FROM Laptop_Sales
);
```

```sql
--6. Find laptops sold on the latest sale date.
SELECT *
FROM Laptop_Sales
WHERE SaleDate =
(
    SELECT MAX(SaleDate)
    FROM Laptop_Sales
);
```

```sql
--7. Find laptops sold after the earliest sale date.
SELECT *
FROM Laptop_Sales
WHERE SaleDate >
(
    SELECT MIN(SaleDate)
    FROM Laptop_Sales
);
```

```sql
--8. Find the brand(s) with the highest total quantity sold.
SELECT Brand
FROM
(
    SELECT Brand, SUM(QuantitySold) AS TotalSold
    FROM Laptop_Sales
    GROUP BY Brand
) 
WHERE TotalSold =
(
    SELECT MAX(TotalSold)
    FROM
    (
        SELECT SUM(QuantitySold) AS TotalSold
        FROM Laptop_Sales
        GROUP BY Brand
    ) 
);
```

```sql
--9. Find laptops that cost more than every Dell laptop.

SELECT *
FROM Laptop_Sales
WHERE Price >
(
    SELECT MAX(Price)
    FROM Laptop_Sales
    WHERE Brand = 'Dell'
);
```

```sql
-- 10. Find laptops whose price matches the price of any Apple laptop.
SELECT *
FROM Laptop_Sales
WHERE Price IN
(
    SELECT Price
    FROM Laptop_Sales
    WHERE Brand = 'Apple'
);
```
