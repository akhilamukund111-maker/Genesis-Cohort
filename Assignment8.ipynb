```sql
--1. Find all policies where the premium is greater than the average premium.
WITH AvgPremium AS
(
    SELECT AVG(Premium) AS Avg_Premium
    FROM MarineInsurance
)
SELECT *
FROM MarineInsurance
WHERE Premium > (SELECT Avg_Premium FROM AvgPremium);
```

```sql

--2. Find the customer(s) with the highest policy amount.
WITH MaxPolicy AS (
    SELECT MAX(PolicyAmount) AS Max_Policy
    FROM MarineInsurance
)
SELECT CustomerName, PolicyAmount
FROM MarineInsurance
WHERE PolicyAmount = (SELECT Max_Policy FROM MaxPolicy);
```

```sql
--3. Find the customer(s) with the highest claim amount.
WITH MaxClaim AS (
    SELECT MAX(ClaimAmount) AS Max_Claim
    FROM MarineInsurance
)
SELECT CustomerName, ClaimAmount
FROM MarineInsurance
WHERE ClaimAmount = (SELECT Max_Claim FROM MaxClaim);

```

```sql
--4. Calculate the total premium collected from each port.
WITH PortPremium AS (
    SELECT Port, SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)
SELECT *
FROM PortPremium;
```

```sql
--5. Find ports where the total premium collected is greater than ₹15,000.
WITH PortPremium AS (
    SELECT Port, SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)
SELECT *
FROM PortPremium
WHERE TotalPremium > 15000;
```

```sql
--6. Calculate the total policy amount for each cargo type.
WITH CargoPolicy AS (
    SELECT CargoType, SUM(PolicyAmount) AS TotalPolicyAmount
    FROM MarineInsurance
    GROUP BY CargoType
)
SELECT *
FROM CargoPolicy;
```

```sql
--7. Find the average policy amount for each port.
WITH PortAverage AS (
    SELECT Port, AVG(PolicyAmount) AS AvgPolicyAmount
    FROM MarineInsurance
    GROUP BY Port
)
SELECT *
FROM PortAverage;
```

```sql
--8. Find policies whose policy amount is greater than the average policy amount.
WITH AvgPolicy AS (
    SELECT AVG(PolicyAmount) AS Avg_Policy
    FROM MarineInsurance
)
SELECT m.*
FROM MarineInsurance m
CROSS JOIN AvgPolicy a
WHERE m.PolicyAmount > a.Avg_Policy;
```

```sql
--9. Find the port with the highest total policy amount.
WITH PortPolicy AS (
    SELECT Port, SUM(PolicyAmount) AS TotalPolicyAmount
    FROM MarineInsurance
    GROUP BY Port
)
SELECT *
FROM PortPolicy
ORDER BY TotalPolicyAmount DESC
LIMIT 1;
```

```sql
--10. Find policies whose claim amount is greater than the average claim amount.
WITH AvgClaim AS (
    SELECT AVG(ClaimAmount) AS Avg_Claim
    FROM MarineInsurance
)
SELECT m.*
FROM MarineInsurance m
CROSS JOIN AvgClaim a
WHERE m.ClaimAmount > a.Avg_Claim;
```

```sql
--11. Find policies where the policy amount is greater than the average policy amount of their own port.
WITH PortAverage AS (
    SELECT
        Port,
        AVG(PolicyAmount) AS AvgPolicyAmount
    FROM MarineInsurance
    GROUP BY Port
)
SELECT
    m.PolicyID,
    m.CustomerName,
    m.Port,
    m.PolicyAmount
FROM MarineInsurance m
JOIN PortAverage p
ON m.Port = p.Port
WHERE m.PolicyAmount > p.AvgPolicyAmount;
```

```sql
--12. Find cargo types whose total claim amount exceeds ₹100,000.
WITH CargoClaim AS (
    SELECT
        CargoType,
        SUM(ClaimAmount) AS TotalClaimAmount
    FROM MarineInsurance
    GROUP BY CargoType
)
SELECT *
FROM CargoClaim
WHERE TotalClaimAmount > 100000;
```

```sql
--13. Find customers whose premium is greater than the average premium of their port.
WITH PortAvgPremium AS (
    SELECT
        Port,
        AVG(Premium) AS AvgPremium
    FROM MarineInsurance
    GROUP BY Port
)
SELECT
    m.CustomerName,
    m.Port,
    m.Premium
FROM MarineInsurance m
JOIN PortAvgPremium p
ON m.Port = p.Port
WHERE m.Premium > p.AvgPremium;
```

```sql
--14. Find policies that contribute more than 25% of their port's total premium.
WITH PortTotalPremium AS (
    SELECT
        Port,
        SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)
SELECT
    m.PolicyID,
    m.CustomerName,
    m.Port,
    m.Premium
FROM MarineInsurance m
JOIN PortTotalPremium p
ON m.Port = p.Port
WHERE m.Premium > (p.TotalPremium * 0.25);
```
