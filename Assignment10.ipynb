```sql
-- 1. Calculate total sales for each Department using SUM() OVER().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    SUM(Sales_Amount) OVER(PARTITION BY Department) AS Department_Total_Sales
FROM gold_sales;
```

```sql
-- 2. Calculate average sales for each Department using AVG() OVER().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    AVG(Sales_Amount) OVER(PARTITION BY Department) AS Department_Avg_Sales
FROM gold_sales;
```

```sql
-- 3. Calculate company-wide total sales using SUM() OVER().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    SUM(Sales_Amount) OVER() AS Company_Total_Sales
FROM gold_sales;
```

```sql
-- 4. Calculate company-wide average sales using AVG() OVER().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    AVG(Sales_Amount) OVER() AS Company_Avg_Sales
FROM gold_sales;
```

```sql
-- 5. Calculate a running total of sales ordered by Order_Date.
SELECT
    Order_ID,
    Customer_Name,
    Order_Date,
    Sales_Amount,
    SUM(Sales_Amount) OVER(
        ORDER BY Order_Date, Order_ID
    ) AS Running_Total
FROM gold_sales;
```

```sql
-- 6. Calculate a running total separately for each Department.
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Order_Date,
    Sales_Amount,
    SUM(Sales_Amount) OVER(
        PARTITION BY Department
        ORDER BY Order_Date, Order_ID
    ) AS Department_Running_Total
FROM gold_sales;
```

```sql
-- 7. Find the highest-selling customer in each Department using ROW_NUMBER().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount
FROM (
    SELECT
        Order_ID,
        Customer_Name,
        Department,
        Sales_Amount,
        ROW_NUMBER() OVER(
            PARTITION BY Department
            ORDER BY Sales_Amount DESC
        ) AS rn
    FROM gold_sales
) 
WHERE rn = 1;
```

```sql
-- 8. Find the top 2 sales levels in each Department using DENSE_RANK().
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount
FROM (
    SELECT
        Order_ID,
        Customer_Name,
        Department,
        Sales_Amount,
        DENSE_RANK() OVER(
            PARTITION BY Department
            ORDER BY Sales_Amount DESC
        ) AS sales_rank
    FROM gold_sales
) 
WHERE sales_rank <= 2;
```
