```sql
--1. Display the customer name, property name, property price, and sale date.
SELECT c.customer_name, p.property_name, p.price , s.sale_date
from Customers c   
inner join Sales s   
ON c.customer_id = s.customer_id
inner join Properties p 
ON s.property_id = p.property_id;
```

```sql
--2. Display every property along with the agent who manages it.
select p.property_name, a.agent_name
from properties p  
inner join agents a  
ON p.agent_id = a.agent_id;


```

```sql
--3. Show customer name, property name,and agent name for every sale.
SELECT c.customer_name, p.property_name, a.agent_name
FROM Sales s
JOIN Customers c
ON s.customer_id = c.customer_id
JOIN Properties p
ON s.property_id = p.property_id
JOIN Agents a
ON p.agent_id = a.agent_id;
```

```sql
--4. Find all customers who purchased a property worth more than $500,000.
SELECT c.customer_name, p.property_name, p.price
FROM Sales s
JOIN Customers c
ON s.customer_id = c.customer_id
JOIN Properties p
ON s.property_id = p.property_id
WHERE p.price > 500000;
```

```sql
--5. Show all properties sold by agent 'Sarah'.
SELECT p.property_name, p.price
FROM Sales s
JOIN Properties p
ON s.property_id = p.property_id
JOIN Agents a
ON p.agent_id = a.agent_id
WHERE a.agent_name = 'Sarah';
```

```sql
--6. Display customers who purchased properties located in the same city where they live.
SELECT c.customer_name, c.city, p.property_name
FROM Sales s
JOIN Customers c
ON s.customer_id = c.customer_id
JOIN Properties p
ON s.property_id = p.property_id
WHERE c.city = p.city;
```

```sql
--7a. Find the total sales value handled by each agent.
SELECT a.agent_name, SUM(p.price) AS total_sales_value
FROM Sales s
JOIN Properties p
ON s.property_id = p.property_id
JOIN Agents a
ON p.agent_id = a.agent_id
GROUP BY a.agent_name;
```

```sql
--7b. Count how many properties each customer purchased.
SELECT c.customer_name, COUNT(s.property_id) AS properties_purchased
FROM Customers c
JOIN Sales s
ON c.customer_id = s.customer_id
GROUP BY c.customer_name;
```

```sql
--8. Find the most expensive property sold by each agent.
SELECT a.agent_name,
       MAX(p.price) AS highest_property_price
FROM Sales s
JOIN Properties p
ON s.property_id = p.property_id
JOIN Agents a
ON p.agent_id = a.agent_id
GROUP BY a.agent_name;
```

```sql
--9. Generate a complete sales report.
SELECT s.sale_id,
       s.sale_date,
       c.customer_name,
       c.city AS customer_city,
       p.property_name,
       p.property_type,
       p.city AS property_city,
       p.price,
       a.agent_name,
       a.office_city
FROM Sales s
JOIN Customers c
ON s.customer_id = c.customer_id
JOIN Properties p
ON s.property_id = p.property_id
JOIN Agents a
ON p.agent_id = a.agent_id
ORDER BY s.sale_date;
```
