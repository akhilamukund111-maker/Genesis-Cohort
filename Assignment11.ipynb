```sql
--1. CURRENT_DATE: Display customer_name, order_date, and today's date.
SELECT
    customer_name,
    order_date,
    CURRENT_DATE() AS today_date
FROM audio_sales;
```

```sql
--2. DATEDIFF : Calculate the number of days between order_date and delivery_date.
SELECT
    customer_name,
    order_date,
    delivery_date,
    DATEDIFF(delivery_date, order_date) AS delivery_days
FROM audio_sales;
```

```sql
--3. DATE_ADD: Find the date that is 7 days after each order_date.
SELECT
    customer_name,
    order_date,
    DATE_ADD(order_date, 7) AS date_after_7_days
FROM audio_sales;
```

```sql

--4. EXTRACT: Extract the year, month, and day from order_date.
SELECT
    customer_name,
    order_date,
    EXTRACT(YEAR FROM order_date) AS year,
    EXTRACT(MONTH FROM order_date) AS month,
    EXTRACT(DAY FROM order_date) AS day
FROM audio_sales;
```

```sql

--5. CASE WHEN:  Create product_type:
SELECT
    customer_name,
    product_name,
    category,
    CASE
        WHEN category = 'AirPods' THEN 'Apple Product'
        WHEN category = 'Earphones' THEN 'Other Brand'
    END AS product_type
FROM audio_sales;
```

```sql
--6. COALESCE: Replace NULL values in discount with 0.
SELECT
    customer_name,
    discount,
    COALESCE(discount, 0) AS final_discount
FROM audio_sales;
```

```sql

--7. COALESCE: Replace NULL values in rating with 0.
SELECT
    customer_name,
    rating,
    COALESCE(rating, 0) AS final_rating
FROM audio_sales;
```

```sql

--8. NULLIF: Use NULLIF to convert a rating of 0 into NULL.
SELECT
    customer_name,
    rating,
    NULLIF(rating, 0) AS new_rating
FROM audio_sales;
```

```sql

--9. NULLIF: Calculate discount percentage using:  discount / price × 100 Use NULLIF to prevent division-by-zero.
SELECT
    customer_name,
    price,
    discount,
    (discount / NULLIF(price, 0)) * 100 AS discount_percentage
FROM audio_sales;
```

```sql

--10. Combined :Find the delivery days and create a category:
--3 days or less → Fast
--4–5 days → Normal
--More than 5 days → Slow
SELECT
    customer_name,
    order_date,
    delivery_date,
    DATEDIFF(delivery_date, order_date) AS delivery_days,
    CASE
        WHEN DATEDIFF(delivery_date, order_date) <= 3 THEN 'Fast'
        WHEN DATEDIFF(delivery_date, order_date) BETWEEN 4 AND 5 THEN 'Normal'
        WHEN DATEDIFF(delivery_date, order_date) > 5 THEN 'Slow'
    END AS delivery_category
FROM audio_sales;
```

```sql

--11. Combined: Display the final discount using COALESCE, calculate sales using price × quantity, and classify sales using CASE WHEN.
SELECT
    customer_name,
    product_name,
    price,
    quantity,
    COALESCE(discount, 0) AS final_discount,
    price * quantity AS sales,
    CASE
        WHEN price * quantity >= 500 THEN 'High Sales'
        WHEN price * quantity >= 200 THEN 'Medium Sales'
        ELSE 'Low Sales'
    END AS sales_category
FROM audio_sales;
```
