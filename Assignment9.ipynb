```sql
-- Q1. Rank all watches by price (descending) using RANK(). Show model_name, price, and the rank.
SELECT
    model_name,
    price,
    RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql
-- Q2. Repeat Q1 using DENSE_RANK(). Add a comment explaining what would differ from RANK() if two watches had the exact same price.
SELECT
    model_name,
    price,
    DENSE_RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql

-- Q3. Assign a unique sequential number to every sale ordered by sale_date (oldest first) using ROW_NUMBER().
SELECT
    sale_id,
    model_name,
    sale_date,
    ROW_NUMBER() OVER (ORDER BY sale_date ASC) AS sale_number
FROM watch_sales;
```

```sql
-- Q4. Within each category, rank watches by units_sold (highest first) using RANK(), resetting for every category (use PARTITION BY).
SELECT
    model_name,
    category,
    units_sold,
    RANK() OVER (
        PARTITION BY category
        ORDER BY units_sold DESC
    ) AS category_rank
FROM watch_sales;
```

```sql

-- Q5. Find the top 3 best-selling watches (by units_sold) in each category using DENSE_RANK() inside a CTE or subquery.
WITH ranked_watches AS (
    SELECT
        model_name,
        category,
        units_sold,
        DENSE_RANK() OVER (
            PARTITION BY category
            ORDER BY units_sold DESC
        ) AS sales_rank
    FROM watch_sales
)
SELECT
    model_name,
    category,
    units_sold,
    sales_rank
FROM ranked_watches
WHERE sales_rank <= 3;
```

```sql

-- Q6. Use ROW_NUMBER() to identify and list only the most recent sale for each brand_name.
WITH latest_sales AS (
    SELECT
        brand_name,
        model_name,
        sale_date,
        units_sold,
        ROW_NUMBER() OVER (
            PARTITION BY brand_name
            ORDER BY sale_date DESC
        ) AS rn
    FROM watch_sales
)
SELECT
    brand_name,
    model_name,
    sale_date,
    units_sold
FROM latest_sales
WHERE rn = 1;
```

```sql

-- Q7.  Use CONCAT() to build a full_name column from customer_first_name and customer_last_name, separated by a space.
SELECT
    CONCAT(customer_first_name, ' ', customer_last_name) AS full_name
FROM watch_sales;
```

```sql

-- Q8.  brand_name has extra leading/trailing spaces (e.g. ' Casio '). Use TRIM() to clean it up.
SELECT TRIM(brand_name) AS cleaned_brand_name 
FROM watch_sales;
```

```sql

-- Q9. Extract the first 4 characters of model_name for every row using SUBSTRING().
SELECT
    model_name,
    SUBSTRING(model_name, 1, 4) AS first_four_characters
FROM watch_sales;
```

```sql

-- Q10. model_name uses hyphens (e.g. 'Submariner-Date'). Use REPLACE() to convert all hyphens to spaces.
SELECT
    model_name,
    REPLACE(model_name, '-', ' ') AS new_model_name
FROM watch_sales;
```

```sql

-- Q11. Standardize the country column to all uppercase using UPPER(). Then write a second version using LOWER().
SELECT country,
    UPPER(country) AS uppercase_country
FROM watch_sales;
```

```sql

-- Q12. Combine TRIM() and UPPER() in a single query to clean and standardize brand_name in one step.
SELECT
    brand_name,
    UPPER(TRIM(brand_name)) AS cleaned_brand_name
FROM watch_sales;
```

```sql

-- Q13. Clean customer_first_name and customer_last_name with TRIM() and proper case (UPPER/LOWER), combine them into full_name with CONCAT(), then rank customers by total units_sold using DENSE_RANK().
WITH customer_sales AS (
    SELECT
        UPPER(TRIM(customer_first_name)) AS first_name,
        UPPER(TRIM(customer_last_name)) AS last_name,
        SUM(units_sold) AS total_units_sold
    FROM watch_sales
    GROUP BY
        UPPER(TRIM(customer_first_name)),
        UPPER(TRIM(customer_last_name))
)
SELECT
    CONCAT(first_name, ' ', last_name) AS full_name,
    total_units_sold,
    DENSE_RANK() OVER (
        ORDER BY total_units_sold DESC
    ) AS sales_rank
FROM customer_sales;
```

```sql

-- Q14. Using SUBSTRING() and CONCAT(), mask customer_last_name so only the first letter shows and the rest is asterisks (e.g. 'KUMAR' -> 'K****'). Then use ROW_NUMBER() to number the masked list alphabetically by customer_last_name.
WITH masked_names AS (
    SELECT
        customer_first_name,
        customer_last_name,
        CONCAT(
            SUBSTRING(customer_last_name, 1, 1),
            REPEAT('*', LENGTH(customer_last_name) - 1)
        ) AS masked_last_name
    FROM watch_sales
)
SELECT
    customer_first_name,
    masked_last_name,
    ROW_NUMBER() OVER (
        ORDER BY customer_last_name ASC
    ) AS row_number
FROM masked_names;
```
