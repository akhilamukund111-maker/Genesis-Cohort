```sql
-- Q3. CASE WHEN
-- Using the sales table, create a column called Sales_Category:
-- Sales_Amount >= 1500  → 'High'
-- Sales_Amount >= 800   → 'Medium'
-- Otherwise             → 'Low'
SELECT
    Sale_ID,
    Customer_Name,
    Product,
    Sales_Amount,
    CASE
        WHEN Sales_Amount >= 1500 THEN 'High'
        WHEN Sales_Amount >= 800 THEN 'Medium'
        ELSE 'Low'
    END AS Sales_Category
FROM sales;
```

```sql
-- Q4. CASE WHEN
-- Create a column called Order_Result:
-- Status = 'Completed' → 'Successful'
-- Status = 'Cancelled' → 'Unsuccessful'
-- Display Sale_ID, Customer_Name, Status and Order_Result.
SELECT
    Sale_ID,
    Customer_Name,
    Status,
    CASE
        WHEN Status = 'Completed' THEN 'Successful'
        WHEN Status = 'Cancelled' THEN 'Unsuccessful'
    END AS Order_Result
FROM sales;
```

```sql
-- Q5. LAG()
-- For each customer, display:
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Previous_Month_Sales
-- Use LAG() to get the previous month's sales.
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales
FROM sales
ORDER BY Customer_Name, Sale_Date;

```

```sql
-- Q6. LEAD()
-- For each customer, display:
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Next_Month_Sales
-- Use LEAD() to get the next month's sales.
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    LEAD(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Next_Month_Sales
FROM sales;

```

```sql
-- Q7. MONTH-TO-MONTH SALES CHANGE
-- Calculate the difference between the current month's
-- sales and the previous month's sales.
-- Formula:
-- Current Sales - Previous Month Sales
-- Use LAG().
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales,
    
    Sales_Amount -
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Sales_Change
FROM sales;
```

```sql
-- Q8. MONTH-TO-MONTH PERCENTAGE
-- Calculate the month-to-month percentage change in sales.
-- Formula:
-- (Current Sales - Previous Sales)
-- / Previous Sales * 100
-- Use LAG().
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales,
    
    (
        (Sales_Amount -
        LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        ))
        /
        LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        )
    ) * 100 AS Percentage_Change

FROM sales;
```

```sql
-- Q9. SALES TREND
-- Using LAG() and CASE WHEN, create a column called Sales_Trend:
-- No previous sale → 'First Month'
-- Current Sales > Previous Sales → 'Increased'
-- Current Sales < Previous Sales → 'Decreased'
-- Current Sales = Previous Sales → 'No Change'
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    
    CASE
        WHEN LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        ) IS NULL
        THEN 'First Month'
        
        WHEN Sales_Amount >
             LAG(Sales_Amount) OVER (
                 PARTITION BY Customer_Name
                 ORDER BY Sale_Date
             )
        THEN 'Increased'
        
        WHEN Sales_Amount <
             LAG(Sales_Amount) OVER (
                 PARTITION BY Customer_Name
                 ORDER BY Sale_Date
             )
        THEN 'Decreased'
        
        ELSE 'No Change'
    END AS Sales_Trend

FROM sales;


```

```sql
-- Q10. LAG() + LEAD()
-- Display the following for each customer:
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Previous_Month_Sales
-- Next_Month_Sales
-- Use both LAG() and LEAD() in the same query.
SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales,
    
    LEAD(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Next_Month_Sales

FROM sales;
```
