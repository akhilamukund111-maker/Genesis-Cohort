```sql
--1. Display employee names and their department names.
SELECT e.emp_name, d.dept_name
FROM Employees e
INNER JOIN Departments d
ON e.dept_id = d.dept_id;
```

```sql
--2. Display employee names along with their department locations.
SELECT e.emp_name, d.location
FROM Employees e
INNER JOIN Departments d
ON e.dept_id = d.dept_id;
```

```sql
--3. Display employee ID, employee name, and department name.
SELECT e.emp_id, e.emp_name, d.dept_name
FROM Employees e
INNER JOIN Departments d
ON e.dept_id = d.dept_id;


```

```sql
--4. Find all employees working in the IT department.
SELECT e.*
FROM Employees e
INNER JOIN Departments d
ON e.dept_id = d.dept_id
WHERE d.dept_name = 'IT';
```

```sql
--5. Find all employees whose department is located in Austin.
SELECT e.*
FROM Employees e
INNER JOIN Departments d
ON e.dept_id = d.dept_id
WHERE d.location = 'Austin';
```

```sql
--6. Display all employees along with their department names, including employees without a matching department.
SELECT e.emp_id, e.emp_name, d.dept_name
FROM Employees e 
LEFT JOIN Departments d   
ON e.dept_id = d.dept_id;
```

```sql
--7. Display all employees and their department locations.
SELECT e.emp_name, d.location
FROM Employees e
LEFT JOIN Departments d
ON e.dept_id = d.dept_id;
```

```sql
--8. Find employees who do not have a matching department.
SELECT e.*
FROM Employees e
LEFT JOIN Departments d
ON e.dept_id = d.dept_id
WHERE d.dept_id IS NULL;
```

```sql
--9. Display employee names and department names. Show NULL where no department exists.
SELECT e.emp_name, d.dept_name
FROM Employees e
LEFT JOIN Departments d
ON e.dept_id = d.dept_id;
```

```sql
--10. Count the number of employees in each department using a LEFT JOIN.
SELECT d.dept_name, COUNT(e.emp_id) AS employee_count
FROM Departments d
LEFT JOIN Employees e
ON d.dept_id = e.dept_id
GROUP BY d.dept_name;
```

```sql
--11. Display all departments along with their employees, including departments with no employees.
SELECT d.dept_name, e.emp_name
FROM Departments d
LEFT JOIN Employees e
ON d.dept_id = e.dept_id;
```

```sql
--12. Find departments that have no employees.
SELECT d.dept_name
FROM Departments d
LEFT JOIN Employees e
ON d.dept_id = e.dept_id
WHERE e.emp_id IS NULL;
```

```sql
--13. Display every department, even if there are no employees assigned to it.

SELECT d.dept_id, d.dept_name, e.emp_name
FROM Departments d
LEFT JOIN Employees e
ON d.dept_id = e.dept_id;
```

```sql
--14. Display all employees and all departments, whether they match or not.
SELECT e.emp_name, d.dept_name
FROM Employees e  
FULL OUTER JOIN Departments d 
ON e.dept_id = d.dept_id;


```

```sql
--15a. Find:Employees without matching departments.
SELECT e.*
FROM Employees e
LEFT JOIN Departments d
ON e.dept_id = d.dept_id
WHERE d.dept_id IS NULL;

```

```sql
--15b. Departments without employees.
SELECT d.*
FROM Departments d
LEFT JOIN Employees e
ON d.dept_id = e.dept_id
WHERE e.emp_id IS NULL;
```
