```sql
--4. Sort customers by loan amount(highest to lowest) 
select * from customers
    ORDER BY Loan_amount DESC;

   
```

```sql
 --5. Show only the top 5 customers based on loan amount
 select * from customers
    ORDER BY Loan_amount DESC
    LIMIT 5;

    
```

```sql
--6. Display the count of customers
    select 
    COUNT(*) AS Total_Customers
    from Customers;

   
```

```sql
--7. Display the total loan amount.
 select
    SUM(Loan_amount) AS Total_Loan_amount
    from Customers;

    
```

```sql

--8. Display the minimum balance.
    select
    MIN(Balance) AS Minimum_Balance
    from Customers;

    
```

```sql
--9. Display the maximum balance.
select
    MAX(Balance) AS Maximim_Balance
    from Customers;
    
   
```

```sql
--10. Display the average balance.
select
    AVG(Balance) AS Average_Balance
    from Customers;
```
